```python
import os
import pandas as pd

def explore_datasets_v2():
    # --- 1. CONFIGURAZIONE ---
    base_path = r"C:\Users\chris\Desktop\0_ULTRABACKUP_DELL\UNISA_Magistrale\1_I anno - I semestre\Programming for data science\Progetto\Testing\files"
    
    # Distinguiamo i file per tipo di separatore atteso
    aifa_files = [
        "atc.csv", "confezioni.csv", "elenco_medicinali_carenti.csv",
        "Lista_farmaci_equivalenti.csv", "PA_confezioni.csv"
    ]
    other_csv_files = ["Emergency_Service_Triage_Application_data.csv"]
    xml_files = ["full database.xml"]
    
    output_dir = "files_samples_v2" # Usiamo una nuova cartella per i nuovi report
    
    # --- 2. CREAZIONE CARTELLA DI OUTPUT ---
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        print(f"Cartella '{output_dir}' creata con successo.")

    # --- 3. PROCESSO DI ANALISI ---
    all_files = aifa_files + other_csv_files + xml_files
    for filename in all_files:
        file_path = os.path.join(base_path, filename)
        report_filename = f"{os.path.splitext(filename)[0]}_sample.txt"
        report_filepath = os.path.join(output_dir, report_filename)

        print(f"Analizzando '{filename}'...")

        try:
            with open(report_filepath, 'w', encoding='utf-8') as report_file:
                report_file.write(f"--- ANTEPRIMA DEL FILE: {filename} ---\n\n")

                if filename in aifa_files or filename in other_csv_files:
                    # --- Gestione dei file CSV con Pandas ---
                    separator = ';' if filename in aifa_files else ','
                    
                    try:
                        # Tentativo con UTF-8
                        df = pd.read_csv(file_path, sep=separator, on_bad_lines='skip', engine='python')
                    except UnicodeDecodeError:
                        # Fallback a latin-1 se UTF-8 fallisce
                        print(f" -> Warning: UTF-8 fallito per {filename}. Tento con latin-1.")
                        df = pd.read_csv(file_path, sep=separator, on_bad_lines='skip', engine='python', encoding='latin-1')
                    except Exception as e:
                        report_file.write(f"Errore imprevisto durante la lettura del CSV: {e}\n")
                        continue

                    report_file.write("Prime 5 righe:\n")
                    report_file.write(df.head().to_string())
                    report_file.write("\n\nInformazioni sulle colonne e tipi di dato:\n")
                    
                    from io import StringIO
                    buffer = StringIO()
                    df.info(buf=buffer)
                    report_file.write(buffer.getvalue())

                elif filename in xml_files:
                    # --- Gestione sicura del file XML ---
                    report_file.write("Prime 50 righe del file XML (lettura come testo):\n\n")
                    with open(file_path, 'r', encoding='utf-8', errors='ignore') as xml_file:
                        for i, line in enumerate(xml_file):
                            if i >= 50:
                                break
                            report_file.write(line)
                
                print(f" -> Report '{report_filename}' creato.")

        except FileNotFoundError:
            print(f" !!! ERRORE: File non trovato: {file_path}")
        except Exception as e:
            print(f" !!! ERRORE: Si è verificato un errore inaspettato con il file {filename}: {e}")

    print(f"\nAnalisi completata. Tutti i report sono stati generati nella cartella '{output_dir}'.")

if __name__ == "__main__":
    explore_datasets_v2()
```
